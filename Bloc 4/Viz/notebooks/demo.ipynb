```python
import panel as pn

pn.extension()

pn.panel("Premiers pas avec Panel !!!!!!").servable()

pn.panel("""
# Mon application

Les étudiants du [Cepe](https://www.lecepe.fr/) sont :

- *motivés*,
- **enthousiastes**,
- <span style="color:red">formidables</span>,
- ...
""").servable()

pn.panel("data/gandalf.jpg", width=256).servable()

pn.panel("/home/onyxia/work/Onyxia/Onyxia/Bloc 4/Viz/notebooks/data/gandalf.mp3").servable()

notes = {
"Nom": ["Teddy", "Jenny", "Cindy", "Franky", "Jimmy"],
"Note": [12, 18, 14, 12, 16],
}

pn.panel(notes).servable()

import pandas as pd

df = pd.DataFrame(notes)

pn.panel(df).servable()

import matplotlib.pyplot as plt

fig, ax = plt.subplots()
ax.bar(df.Nom, df.Note)
ax.grid(linestyle=":")

plt.close(fig) # Important pour éviter les fuites de mémoire

pn.panel(fig, width=512).servable()

import plotly.express as px

pn.extension("plotly") # Nécessaire pour Plotly

fig_plotly = px.line(df, x="Nom", y="Note")
fig_plotly.update_traces(mode="lines+markers")
fig_plotly.layout.autosize = True

pn.panel(fig_plotly).servable()
```
